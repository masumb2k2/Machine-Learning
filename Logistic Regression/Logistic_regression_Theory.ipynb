# Logistic Regression

```python
# import modules
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
# Random DataFrame Generation
np.random.seed(0)
df=pd.DataFrame(
data={
    'Feature': np.arange(1,101),
    'Target':np.array([0]*60 + [1]*40)
}
)
df.head()
```

```python
# linear  model plot
sns.lmplot(data=df,x='Feature',y='Target')
```

```python
# Noise addition to DataFrame
df_jitter=df.copy()
df_jitter['Target']=df_jitter['Target']+np.random.normal(0,0.3,size=len(df))
df_jitter['Target']=df_jitter['Target'].clip(0,1)
```

```python
df_jitter.head()
```

```python
# Regression plot with jittered data
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))
sns.regplot(
    data=df_jitter,
    x='Feature',
    y='Target',
    label='Linear Regression',
)
sns.regplot(
    data=df_jitter,
    x='Feature',
    y='Target',
    logistic=True,
    color='red',
)
plt.legend()
plt.show()
```
